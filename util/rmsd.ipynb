```python
def parse_and_align(pdb1, pdb2):
    """
    Computes rmsd between two pdb files on the backbone. 
    """

    parsed1 = parsers.parse_pdb(pdb1)
    xyz1 = parsed1['xyz'][:,:3,:]     # only N-CA-C

    parsed2 = parsers.parse_pdb(pdb2)
    xyz2 = parsed2['xyz'][:,:3,:]     # only N-CA-C


    rmsd, _, _ = kinematics.np_kabsch(xyz1.reshape(-1,3), xyz2.reshape(-1,3))

    return rmsd 
```
